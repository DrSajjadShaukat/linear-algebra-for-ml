## 10.10  Python implementation

NumPy returns U, a one-dimensional array of singular values, and Vh. For our real matrices, Vh is Vᵀ. Its rows are right singular vectors. Reduced shapes are obtained with full_matrices=False.

```python
import numpy as np
A = np.array([[3., 0.], [4., 0.], [0., 2.]])
U, s, Vh = np.linalg.svd(A, full_matrices=False)
print(U.shape, s.shape, Vh.shape)
print(s)  # [5. 2.]
rebuilt = (U * s) @ Vh
assert np.allclose(rebuilt, A)
assert np.allclose(U.T @ U, np.eye(2))
assert np.allclose(Vh @ Vh.T, np.eye(2))
k = 1
Ak = (U[:, :k] * s[:k]) @ Vh[:k, :]
error = np.sum((A - Ak)**2)
assert np.isclose(error, np.sum(s[k:]**2))
```

U * s scales U’s columns. The reduced shapes are (3, 2), (2,), and (2, 2). Keeping one term gives squared error 4, the square of the discarded singular value 2. Do not write U @ s @ Vh: s is an array, not the diagonal matrix Σ.

### Check a least squares result

```python
b = np.array([4., 3., 4.])
x = np.linalg.pinv(A) @ b
residual = b - A @ x
assert np.allclose(x, [24/25, 2])
assert np.allclose(A.T @ residual, 0)
assert np.isclose(residual @ residual, 49/25)
```

For fitting a system, np.linalg.lstsq is also appropriate. A pseudoinverse is useful here to expose the mathematics; it need not be formed explicitly every time a least squares problem is solved. Numerical routines apply rank tolerances, so near-zero values require interpretation.

## 10.13  Optional extension: Matrix norms and singular values

### A matrix penalty in a model

If a model uses a weight matrix W rather than one coefficient vector, a squared Frobenius penalty λ‖W‖F² equals the sum of squared penalties on every entry. It is the matrix counterpart of squared L₂ regularization. A spectral-norm penalty instead targets maximum amplification and is a different optimization choice.

```python
A = np.diag([3., 4.])
assert np.isclose(np.linalg.norm(A, 'fro'), 5.)
assert np.isclose(np.linalg.norm(A, 2), 4.)
dx = np.array([.1, -.2])
assert np.linalg.norm(A @ dx) <= 4*np.linalg.norm(dx)
W = np.array([[1., -2.], [0., 2.]])
assert np.isclose(np.linalg.norm(W, 'fro')**2, 9.)
```
