## 8.9  Python implementation and interpretation

The following code repeats Example 8. NumPy arrays hold the observations; column_stack constructs the ones column and feature column. The solver returns coefficients, residual information, numerical rank, and singular values.

```python
import numpy as np
x = np.array([0., 1., 2.])
y = np.array([1., 2., 4.])
X = np.column_stack([np.ones(len(x)), x])
beta, reported, rank, singular = np.linalg.lstsq(
    X, y, rcond=None
)
pred = X @ beta
residual = y - pred
sse = residual @ residual
print(beta, pred, sse, rank)
assert np.allclose(X.T @ residual, 0)
```

Expected coefficients are approximately (0.833333, 1.5); predictions are (0.833333, 2.333333, 3.833333); SSE is about 0.166667; rank is 2. Compute SSE from the residual explicitly because the solver’s returned residual array can be empty for some matrix shapes or ranks.

### Check QR against the direct solver

```python
Q, R = np.linalg.qr(X, mode="reduced")
beta_qr = np.linalg.solve(R, Q.T @ y)
assert np.allclose(beta_qr, beta)
```

This simple QR solve assumes full column rank. It is appropriate for our small independent-column example; it is not a replacement for rank handling in general.

## 8.14  Optional extension: Lasso and the meaning of zero

### Example 20  Solve a one-variable Lasso problem

This scalar formula is not a general multivariable Lasso solver. With interacting columns, iterative optimization is needed; applying the formula independently is valid only under suitable orthogonality conditions.

```python
x = np.array([-1., 0., 1.])
y = np.array([-2., 0., 2.])
a, c = x @ x, x @ y
ridge = c / (a + 2.)
lasso = np.sign(c) * max(abs(c) - 4./2, 0) / a
assert np.isclose(ridge, 1.)
assert np.isclose(lasso, 1.)
```
