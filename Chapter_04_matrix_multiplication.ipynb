## 4.8  Matrix products in Python

Use @ for the matrix product of two-dimensional NumPy arrays. Check shapes explicitly before interpreting the result.

```python
import numpy as np
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 3]])
print(A @ B)
# [[ 4  6]
#  [10 12]]
print(B @ A)
# [[ 2  4]
#  [10 14]]
print(A * B)
# [[ 2  0]
#  [ 3 12]]
```

### A transparent implementation

The loops below reproduce the row-by-column rule. They are for learning; use NumPy’s matrix multiplication for numerical work.

```python
def matmul_by_hand(A, B):
    A = np.asarray(A, dtype=float)
    B = np.asarray(B, dtype=float)
    if A.ndim != 2 or B.ndim != 2:
        raise ValueError("Expected two matrices")
    if A.shape[1] != B.shape[0]:
        raise ValueError("Inner dimensions must match")
    C = np.zeros((A.shape[0], B.shape[1]))
    for i in range(A.shape[0]):
        for j in range(B.shape[1]):
            for k in range(A.shape[1]):
                C[i, j] += A[i, k] * B[k, j]
    return C

assert np.allclose(matmul_by_hand(A, B), A @ B)
```

### Calculate several outputs

```python
X = np.array([[2.,3.],[1.5,2.],[2.5,3.]])
W = np.array([[80.,70.],[20.,30.]])
bias = np.array([50.,40.])
Y = X @ W + bias
print(Y.shape)  # (3, 2)
```
