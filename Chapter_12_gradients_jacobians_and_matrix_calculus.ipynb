## 12.5  Python checks and practice

```python
import numpy as np
X = np.array([[1., 0.], [0., 1.], [1., 1.]])
y = np.array([1., 2., 3.])
def loss(beta):
    return np.sum((X @ beta - y)**2)
def grad(beta):
    return 2 * X.T @ (X @ beta - y)
beta = np.array([.3, -.2])
eps = 1e-6
numeric = np.zeros_like(beta)
for j in range(len(beta)):
    d = np.zeros_like(beta)
    d[j] = eps
    numeric[j] = (loss(beta+d) - loss(beta-d))/(2*eps)
assert np.allclose(grad(beta), numeric, atol=1e-6)
beta = np.zeros(2)
for step in range(200):
    beta -= .1 * grad(beta)
assert np.allclose(beta, [1., 2.])
```
