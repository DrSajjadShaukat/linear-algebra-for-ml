## 1.5  Verify the calculations with Python

### A direct calculation

```python
x = [2.0, 3.0]
w = [80.0, 20.0]
b = 50.0
prediction = w[0] * x[0] + w[1] * x[1] + b
print(prediction)  # 270.0
```

A list stores the feature values. The two products are the contributions from area and bedrooms. Their sum plus b is expressed in thousands of dollars.

### A reusable calculation

```python
def predict(features, weights, intercept):
    if len(features) != len(weights):
        raise ValueError("Feature and weight counts must match")
    total = sum(a * b for a, b in zip(features, weights))
    return total + intercept

print(predict([1.5, 2], [80, 20], 50))  # 210.0
```

zip pairs matching entries. sum adds the products. The length check matters because zip otherwise stops at the shorter list. Equal lengths still do not guarantee correct feature order; that is our responsibility.

### The same operation in NumPy

```python
import numpy as np
x = np.array([2.0, 3.0])
w = np.array([80.0, 20.0])
print(w * x)       # [160.  60.]
print(w @ x)       # 220.0
print(w @ x + 50)  # 270.0
```
