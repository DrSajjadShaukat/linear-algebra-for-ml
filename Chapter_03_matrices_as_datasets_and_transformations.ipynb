## 3.8  Matrix shapes in Python

Use NumPy arrays to store numerical matrices. A list of rows creates a two-dimensional array. Python indexing starts at zero, so mathematical entry x₂₁ is X[1, 0].

```python
import numpy as np
X = np.array([[2.0, 3.0],
              [1.5, 2.0],
              [2.5, 3.0]])
print(X.shape)  # (3, 2)
print(X[1, 0])  # 1.5
print(X[:, 0])  # [2.  1.5 2.5]
print(X.T.shape)  # (2, 3)
```

### One-dimensional and column arrays

```python
w = np.array([80.0, 20.0])
w_col = w.reshape(2, 1)
print(w.shape)      # (2,)
print(w.T.shape)    # (2,)
print(w_col.shape)  # (2, 1)
print(w_col.T.shape)  # (1, 2)
```

Transposing a one-dimensional array does not turn it into a column. Reshaping explicitly creates that column. Keep this distinction in mind when comparing NumPy output with the book’s column-vector notation.

### Predictions and elementwise multiplication

```python
print(X @ w + 50)  # [270. 210. 310.]
print((X @ w_col + 50).shape)  # (3, 1)
print(X * w)
# [[160.  60.]
#  [120.  40.]
#  [200.  60.]]
```
