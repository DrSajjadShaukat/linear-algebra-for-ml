## 2.7  Geometry in Python

The companion notebook contains these code blocks, ready to run. NumPy arrays perform vector addition and subtraction componentwise. Ordinary Python lists behave differently: [1, 2] + [3, 4] concatenates two lists.

```python
import numpy as np
a = np.array([2.0, 1.0])
b = np.array([1.0, 2.0])
print(a + b)              # [3. 3.]
print(a - b)              # [ 1. -1.]
print(2 * a)              # [4. 2.]
print(np.linalg.norm([3, 4]))  # 5.0
```

### Distance and a unit vector

```python
p = np.array([1.0, 2.0])
q = np.array([4.0, 6.0])
print(np.linalg.norm(q - p))  # 5.0

x = np.array([3.0, 4.0])
length = np.linalg.norm(x)
if length == 0:
    raise ValueError("Zero vector has no unit direction")
u = x / length
print(u)  # [0.6 0.8]
```

### Angle between nonzero vectors

```python
a = np.array([1.0, 0.0])
b = np.array([1.0, 1.0])
na, nb = np.linalg.norm(a), np.linalg.norm(b)
if na == 0 or nb == 0:
    raise ValueError("Angle requires nonzero vectors")
cosine = (a @ b) / (na * nb)
cosine = np.clip(cosine, -1.0, 1.0)
angle = np.degrees(np.arccos(cosine))
print(round(angle, 2))  # 45.0
```
