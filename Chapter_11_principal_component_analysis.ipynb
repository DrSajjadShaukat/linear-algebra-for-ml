## 11.9  Python workflow for the hand example

The following code fits unscaled PCA on the four training observations. Rows are observations and columns are features. SVD returns directions as rows of Vh, so transpose the retained rows to obtain the columns of W.

```python
import numpy as np
X = np.array([[13., 23.], [7., 17.],
              [11., 19.], [9., 21.]])
mean = X.mean(axis=0)
Xc = X - mean
U, s, Vh = np.linalg.svd(Xc, full_matrices=False)
variance = s**2 / (len(X) - 1)
ratio = variance / variance.sum()
k = 1
W = Vh[:k].T
Z = Xc @ W
Xhat = Z @ W.T + mean
assert np.allclose(variance, [12., 4/3])
assert np.allclose(ratio, [.9, .1])
assert np.isclose(np.sum((X-Xhat)**2), 4)
```

### Transform held-out rows with the same basis

```python
Xnew = np.array([[12., 20.], [8., 20.]])
Znew = (Xnew - mean) @ W
Xnew_hat = Znew @ W.T + mean
assert np.allclose(Xnew_hat, [[11, 21], [9, 19]])
```
