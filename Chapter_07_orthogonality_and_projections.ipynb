## 7.9  Python checks

Use floating-point arrays and compare approximately. The small examples below verify the hand calculations and the defining geometric properties.

```python
import numpy as np

def project_line(b, u):
    b = np.asarray(b, dtype=float)
    u = np.asarray(u, dtype=float)
    if b.ndim != 1 or b.shape != u.shape:
        raise ValueError("Expected matching vectors")
    denominator = u @ u
    if denominator == 0:
        raise ValueError("Direction must be nonzero")
    return ((u @ b) / denominator) * u

b = np.array([4., 2.])
u = np.array([1., 1.])
p = project_line(b, u)
r = b - p
print(p, r)  # [3. 3.] [ 1. -1.]
assert np.isclose(u @ r, 0)
assert np.isclose(b @ b, p @ p + r @ r)
```

### Project onto the tilted plane

```python
Q = np.array([[1/np.sqrt(2), 0.],
              [1/np.sqrt(2), 0.],
              [0., 1.]])
b = np.array([4., 2., 5.])
assert np.allclose(Q.T @ Q, np.eye(2))
p = Q @ (Q.T @ b)
assert np.allclose(p, [3., 3., 5.])
assert np.allclose(Q.T @ (b - p), 0)
```

### QR produces a suitable basis

```python
A = np.array([[1., 1.], [1., 0.]])
Q, R = np.linalg.qr(A, mode="reduced")
assert np.allclose(Q.T @ Q, np.eye(2))
assert np.allclose(Q @ R, A)
```
