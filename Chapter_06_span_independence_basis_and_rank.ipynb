## 6.8  Check the concepts in Python

Use a numerical solver to recover basis coordinates, and matrix multiplication to verify the result. We use small exact examples here; floating-point outputs still require approximate comparisons.

```python
import numpy as np
B = np.array([[1., 1.], [1., -1.]])
w = np.array([3., 1.])
coordinates = np.linalg.solve(B, w)
print(coordinates)  # [2. 1.]
assert np.allclose(B @ coordinates, w)
```

### Rank and a known null-space direction

```python
A = np.array([[1.,0.,1.], [0.,1.,1.], [1.,1.,2.]])
z = np.array([-1., -1., 1.])
print(np.linalg.matrix_rank(A))  # 2
print(A @ z)                    # [0. 0. 0.]
for t in [-2., 0., 1., 3.]:
    x = np.array([2., 3., 0.]) + t * z
    assert np.allclose(A @ x, [2., 3., 5.])
```

### Compare predictions with different coefficients

```python
X = np.array([[1.,1.,60.],
              [1.,2.,120.],
              [1.,3.,180.]])
beta1 = np.array([40., 10., 0.])
beta2 = np.array([40., 4., 0.1])
assert np.linalg.matrix_rank(X) == 2
assert np.allclose(X @ beta1, X @ beta2)
print(X @ beta1)  # [50. 60. 70.]
```
